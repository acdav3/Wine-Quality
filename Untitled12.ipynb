```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
import statsmodels.api as sm
from statsmodels.stats.outliers_influence import variance_inflation_factor
```

```python
red = pd.read_csv("winequality-red.csv", sep=";")
white = pd.read_csv("winequality-white.csv", sep=";")

red["wine_type"] = "red"
white["wine_type"] = "white"

wine = pd.concat([red, white], ignore_index=True)

wine
```

```python
for col in red.columns[:-1]:  # skip wine_type
    plt.figure()
    sns.histplot(data=wine, x=col, hue="wine_type", kde=True)
    plt.title(f"Histogram of {col} by Wine Type")
    plt.tight_layout()
    plt.show()
```

```python
plt.figure()
sns.boxplot(data=red, x="quality", y="alcohol")
plt.title("Red Wine: Alcohol vs Quality")
plt.show()

plt.figure()
sns.boxplot(data=white, x="quality", y="alcohol")
plt.title("White Wine: Alcohol vs Quality")
plt.show()
```

```python
def prepare_data(df):
    X = df.drop(columns=["quality", "wine_type"])
    y = df["quality"]
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)
    X_scaled = pd.DataFrame(X_scaled, columns=X.columns)
    X_scaled["Intercept"] = 1
    return X_scaled, y
```

```python
X_red, y_red = prepare_data(red)
X_white, y_white = prepare_data(white)

model_red = sm.OLS(y_red, X_red).fit()
model_white = sm.OLS(y_white, X_white).fit()

print("Red wine regression summary")
print(model_red.summary())

print("White wine regression summary")
print(model_white.summary())
```

```python
def calculate_vif(X):
    vif_data = pd.DataFrame()
    vif_data["feature"] = X.columns
    vif_data["VIF"] = [variance_inflation_factor(X.values, i) for i in range(len(X.columns))]
    return vif_data

print("Red wine VIF")
print(calculate_vif(X_red))

print("White wine VIF")
print(calculate_vif(X_white))
```
